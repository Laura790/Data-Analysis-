# Seaborn 
* It is a Python library for data visualization built on top of Matplotlib
* It has more than 100 plotting functions

```python
#Importing libraries 

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

```

```python
# Create a DataFrame
data = {'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
        'Age': [25, 30, 22, 35, 28],
        'City': ['New York', 'London', 'Paris', 'Tokyo', 'Sydney'],
        'Score': [85, 92, 78, 95, 88]}
df_school = pd.DataFrame(data)

# Display the DataFrame
display(df_school)
```

```python
df_school.info()
```

 ### Scatter Plot 

```python
sns.scatterplot(data = df_school, x="Age", y="Score" )
plt.title("Scatter Plot of Age Against Score")
plt.grid(True)
plt.show()
```

### Histogram

```python
sns.histplot(df_school["Score"], bins=5)
plt.title("Histogram of Scores")
plt.grid(True)
plt.show()
```

### Bar Plot


```python
sns.barplot(data=df_school, x= "City", y= "Score")
plt.title("A barplot of Score By City")
plt.xticks(rotation=25, ha="right")
plt.grid(True)
plt.show()
```

### Line Plot

```python
sns.lineplot(data=df_school, x="Age", y="Score")
plt.title(" A lineplot of Age Vs Score")
plt.grid(True)
plt.show()
```

### A Heatmap
* Used to show Corelation 

```python
#Extract all the numbers 
heatmap_numbers= df_school.set_index("Name")[["Age","Score"]]
sns.heatmap(heatmap_numbers, annot=True, cmap="magma")
plt.title("A heatmap of Age Vs Score")
plt.show()
```

# Using Real Data in Data Visualization 

```python
my_dataset= pd.read_csv("sales_data_sample.csv" , encoding= "latin-1")
# Inspection
my_dataset.head()
```

```python
my_dataset.info()
```

```python
# Checking the sum of  missing values in the dataset
my_dataset.isnull().sum()
```

```python
# Dropping the missing values(irrelevant for analysis)
my_dataset = my_dataset.drop(columns=["ADDRESSLINE2", "STATE", "POSTALCODE", "TERRITORY"])

# Inspect  if columns are dropped
my_dataset.isnull().sum()
```

# Data Visualization 

```python
# Distribution of Sales 
plt.figure(figsize=(10,6))
sns.histplot(my_dataset["SALES"],kde="true")
plt.title("Distribution of Sales")
plt.grid(True)
plt.savefig("Distribution of Sales.png")
plt.show()
```

### Summary
* The plot shows the distribution of sales with the count of each sale
* The **highest sales** count is at **2500** sales
* The **lowest sales** count is at **14000** sales 

```python
# Sales by Country
#Define before plotting
sales_by_country=my_dataset.groupby("COUNTRY")["SALES"].sum().sort_values(ascending=False)
plt.figure(figsize=(20,10))
plt.bar(sales_by_country.index, sales_by_country.values)
plt.xlabel("Countries")
plt.ylabel("Total Sales")
plt.title("Sales by Country")
plt.xticks(rotation=45, ha="right")
plt.grid()
plt.savefig("Sales by Country.png")
plt.show()
```

### Summary
* This is a plot for total sales per 19 countries .
* USA has the highest number of total sales
* Ireland has the lowest number of total sales 

```python
# Average Sales by Dealsize
deal_size = my_dataset.groupby("DEALSIZE")["SALES"].mean().sort_values(ascending=False)
plt.figure(figsize=(20,10))
plt.bar(deal_size.index, deal_size.values)
plt.xlabel("Deal sizes")
plt.ylabel("Average Sales")
plt.title("Average Sales by Deal Size")
plt.xticks(rotation=45, ha="right")
plt.grid()
plt.savefig("Average Sales by Deal Size.png")
plt.show()
```

### Summary
* This is the plot of the average sales by Dealsize
* The **Large** Deal size has the highest amount of sales, at **8500**
* The **Small** Deal size has the lowest amount of sales,at  **2100** 

```python
#Convert OrderDate from string to DateTime 

my_dataset['ORDERDATE'] = pd.to_datetime(my_dataset['ORDERDATE'], errors='coerce')

# Extract useful time-based features from 'ORDERDATE' to facilitate analysis.
my_dataset['ORDER_YEAR'] = my_dataset['ORDERDATE'].dt.year
my_dataset['ORDER_MONTH'] = my_dataset['ORDERDATE'].dt.month
my_dataset['ORDER_MONTH_YEAR'] = my_dataset['ORDERDATE'].dt.to_period('M')
```

```python
my_dataset.info()
```

```python
# Plotting of all orders
# Assigning Data
sales_by_month= my_dataset.groupby("ORDER_MONTH_YEAR")["SALES"].sum().reset_index()
sales_by_month["ORDER_MONTH_YEAR"]= sales_by_month["ORDER_MONTH_YEAR"].astype(str)
# Plotting the graph
plt.figure(figsize=(16,7))
sns.lineplot(data=sales_by_month, x="ORDER_MONTH_YEAR", y="SALES", marker="o", color="purple")
plt.title("Yearly  Sales per Month")
plt.xlabel("Month Year")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.grid(True)
# Saving the plotted graph as image
plt.savefig("yearly_sales_per_month.png")
plt.show()
```

### Summary 
* The plot shows yearly sales for every month
* The **highest sales** were recorded in **November 2003** and **November 2004**.
* The **lowest sales** were recorded in **January 2003**.
