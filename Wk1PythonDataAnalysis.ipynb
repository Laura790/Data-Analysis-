# Indexing, Slicing and Iterations 

```python
# Indexing -allows you to access individual elements in a sequence using their position/index. It starts at zero
# There is negative indexing-It starts at -1 from the end of a sequence, coming to the start.

animals = ["Cat", "Lion", "Dog", "Giraffe", "Gazelle"]
print(animals[1])
print(animals[-1])
```

```python
#Slicing-  Allows you to extract a section of a sequence or a subsection of a sequence.
#It has : Starting point(included in slicing) , stopping point(not included in slicing) and then finally step(it is optional)

animals = ["Cat", "Lion", "Dog", "Giraffe", "Gazelle"]
print(animals[1:3]) # The starting point included and the element before the stopping point
print(animals[-2:]) # if not specify the stopping point, everything gets included 
```

```python
print(animals[::-1]) #double colon means reverse from ending point to starting point
```

```python
#Iterations - looping through each element in a sequence.
animals = ["Cat", "Lion", "Dog", "Giraffe", "Gazelle"]
for animal in animals:# for one in many 
    print(animal)     #Each animal printed in its own line 
```

```python
# (Quiz3) 2. Write a program to find the sum of all even numbers in a list (e.g., `[2, 5, 8, 10]`).
nums = [2,5,8,10]
even_sum=0
for num in nums:
    if num % 2 ==0:
        even_sum+= num

print("Sum of even numbers: ",even_sum
     )
```

# Sets and Nested Data Structures 

```python
# Sets-A set is an unordered, mutable and unindexed collection of unique elements. 
my_list= {1,2,3,4,5,3,5,9,10}
print(my_list)
```

```python
# Set Operations
a={1,2,3,7}
b={3,4,5,6}
print(a | b) #this is the union between a and b 
```

```python
print(a & b) # the & gives the intersection(the value shared by two sets)
```

```python
print(a - b) # The values in a but not in b  
```

```python
# Nested Data Structures -It is a dictionary within a dictionary
students =[
    {"name": "Laura",
     "age": 56,
     "course": "Medicine"},
    
    {"name": "Steve",
     "age":29,
     "course": "Accounting"}
     
]
print(students[0]["course"]) # specific value of the course 
```
