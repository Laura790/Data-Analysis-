```python
# Data Cleaning
print(my_dataset.isnull().sum())
```

#  **Calculations**

```python
# SALES BY COUNTRIES
print(my_dataset.groupby ('COUNTRY')['SALES'].sum().sort_values(ascending=False))
```

### A summary of SALES BY COUNTRY 
* There are **19 countries**

* The **Ireland** records the lowest number of SALES by **57756.43**

```python
# Group Dealsize by mean 
average_sale_by_dealsize = my_dataset.groupby("DEALSIZE")["SALES"].mean()
average_sale_by_dealsize
```

### A summary of DealSize Analysis

*  The medium DEALSIZE has a moderate average number of SALES with **4398.43**

# **Numpy**

```python
# Import libraries 
import pandas as pd
import numpy as np 
```

```python
# Create a NumPy array
data = np.array([['Alice', 25, 'Nairobi', 85],
                 ['Bob', 30, 'Lagos', 92],
                 ['Charlie', 22, 'Juba', 78],
                 ['David', 35, 'Accra', 95],
                 ['Eve', 28, 'New York', 88]])

# Create column names
columns = ['Name', 'Age', 'City', 'Score']

# Create a Pandas DataFrame from the NumPy array
my_df = pd.DataFrame(data=data, columns=columns)
```

```python
my_df.head()
```

# **Data Analysis with Numpy**

```python
# Converting values of Ages to integers/numbers 
# Converting values of Scores to integers/numbers 

ages = my_df['Age'].to_numpy(dtype=int)
scores = my_df['Score'].to_numpy(dtype=int)
```

```python
# Calculating the average score 
average_score = np.mean(scores)
average_score
```

```python
#Calculating the Max Score 
max_score = np.max(scores)
max_score
```

```python
#Calculating the Min Score 
min_score = np.min(scores)
min_score
```

```python
# Calculating the Median Score
median_score = np.median(scores)
median_score
```

```python
# Calculating the Standard Deviation
std_score = np.std(scores)
print(f"Standard Deviation:{std_score:.2f}")
```

```python
# Calculating Total Scores
sum_scores = np.sum(scores)
sum_scores
```

### **A Summary of Numpy Analysis**
* The **average** Scores is **87.6**
* The **Max** Score is **95**
* The **Min** Score is **78**
* The **Median** Score is **88**
* The **Standard Deviation** Score is **5.89**
* The **Sum total** of Score is **438**
