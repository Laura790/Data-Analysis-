# Pandas 


```python
import pandas as pd 
import numpy as np
```

```python

# Create a DataFrame manually
data = {'Name': ['Mary', 'Bob', 'Charlie', 'David', 'Eve'],
        'Age': [25, 30, 22, 35, 28],
        'City': ['Nairobi', 'Lagos', 'Juba', 'Accra', 'New York'],
        'Score': [85, 92, 78, 95, 88]}

#Converting a Data to a Dataframe
my_data =pd.DataFrame(data)
print(my_data)
```

```python
my_data.head() #view first 5 elements of the dataframe
# print(my_data.head())
# display(my_data.head())
```

```python
my_data.tail(3) # prints the last 5 elements by default
```

```python
# It gives general info abt the dataframe 
# 1. The columns and rows 
# 2. The data types
# 3.The number/quantity  of data -missing data
# 

my_data.info()
```

```python
# Describe - The statistical bit about a dataset
# my_data.describe()
print(my_data.describe())
```

```python
# counting the missing data in a dataset and find the sum 
my_data.isnull().sum()
```

```python
# unique values 
print(my_data['City'].unique())

#Count the unique values 
print(my_data['City'].value_counts())
```
